# Pig Tracking - DYTB Method
Ăp dá»¥ng phÆ°Æ¡ng phĂ¡p tá»« bĂ i bĂ¡o **DYTB (Dual-YOLOX-Tiny-ByteTrack)**
*(Li et al., Smart Agricultural Technology, 2024 â€” DOI: 10.1016/j.atech.2024.100383)*

### Cáº£i tiáº¿n so vá»›i code gá»‘c:
1. **DYTB Multi-channel Attention** â€” CBAM (Channel + Spatial) attention sau backbone, báº¯t chÆ°á»›c cÆ¡ cháº¿ attention cá»§a Dual-YOLOX-Tiny
2. **Trajectory Interpolation** â€” Ná»™i suy tuyáº¿n tĂ­nh tá»a Ä‘á»™ khi track bá»‹ máº¥t táº¡m thá»i (giáº£m ID switch)
3. **3rd Association: Centre-Distance** â€” ThĂªm vĂ²ng ghĂ©p thá»© 3 dĂ¹ng khoáº£ng cĂ¡ch tĂ¢m thay vĂ¬ chá»‰ IoU thuáº§n
4. **Aspect-ratio aware ReID** â€” Tá»‰ lá»‡ bbox lá»£n lĂ  2:1 (rá»™ng:cao) thay vĂ¬ 1:2 cá»§a ngÆ°á»i
5. **Occlusion-aware Kalman** â€” Äiá»u chá»‰nh noise khi lá»£n bá»‹ che khuáº¥t
6. **Behavior logging** â€” Ghi hĂ nh vi (lie/stand/eat/other) náº¿u model cĂ³ multi-class

## Cell 3 â€” HĂ m tiá»‡n Ă­ch: Mask / ROI / Histogram

```python
# ============================================================
#  MASK / ROI
# ============================================================
def load_mask(path, W, H):
    if not path or not os.path.exists(path):
        return None
    m = cv2.imread(path, cv2.IMREAD_GRAYSCALE)
    if m is None: return None
    if ROI_DILATE_PX > 0:
        ker = cv2.getStructuringElement(cv2.MORPH_ELLIPSE,
                                        (ROI_DILATE_PX, ROI_DILATE_PX))
        m = cv2.dilate(m, ker, 1)
    return cv2.resize(m, (W, H), interpolation=cv2.INTER_NEAREST)

def apply_visual_mask(frame, mask):
    if mask is None: return frame
    m3 = cv2.merge([mask, mask, mask]) if len(mask.shape)==2 else mask
    return cv2.bitwise_and(frame, m3)

def roi_keep(mask, box):
    if mask is None: return True
    x1,y1,x2,y2 = map(int, box)
    H, W = mask.shape[:2]
    x1=max(0,min(W-1,x1)); x2=max(0,min(W-1,x2))
    y1=max(0,min(H-1,y1)); y2=max(0,min(H-1,y2))
    if x2<=x1 or y2<=y1: return False
    if ROI_MODE=="center":
        cx=(x1+x2)//2; cy=(y1+y2)//2
        return mask[cy, cx]==255
    roi = mask[y1:y2, x1:x2]
    if roi.size==0: return False
    return np.count_nonzero(roi==255)/float(roi.size) >= ROI_MIN_COVER

# ============================================================
#  APPEARANCE â€” HSV histogram vá»›i pig-ratio 2:1 (DYTB)
# ============================================================
def extract_hist_hsv(frame, box):
    x1,y1,x2,y2 = map(int, box)
    H, W = frame.shape[:2]
    x1=max(0,min(W-1,x1)); x2=max(0,min(W-1,x2))
    y1=max(0,min(H-1,y1)); y2=max(0,min(H-1,y2))
    empty = np.ones((H_BINS*S_BINS*V_BINS,), np.float32)/(H_BINS*S_BINS*V_BINS)
    if x2<=x1 or y2<=y1: return empty
    crop = frame[y1:y2, x1:x2]
    if crop.size==0: return empty
    # APP_W=96, APP_H=48 â†’ tá»‰ lá»‡ 2:1 phĂ¹ há»£p lá»£n (DYTB)
    crop = cv2.resize(crop, (APP_W, APP_H))
    hsv  = cv2.cvtColor(crop, cv2.COLOR_BGR2HSV)
    hist = cv2.calcHist([hsv],[0,1,2],None,
                        [H_BINS,S_BINS,V_BINS],
                        [0,180,0,256,0,256]).astype(np.float32)
    hist /= (hist.sum() + 1e-6)
    return hist.flatten()

def bhattacharyya(h1, h2):
    return np.clip(1.0 - np.sum(np.sqrt(h1*h2)), 0.0, 1.0)
```

## Cell 4 â€” IDBank + TrajectoryInterpolation (DYTB má»›i)

```python
# ============================================================
#  ID Bank â€” lÆ°u appearance histogram theo ID cá»‘ Ä‘á»‹nh
# ============================================================
class IDBank:
    def __init__(self, id_pool, bank_size=60):
        self.bank_size = bank_size
        self.id_pool   = sorted(id_pool)
        self.bank = {fid: collections.deque(maxlen=bank_size)
                     for fid in self.id_pool}

    def update(self, fid, hist):
        if fid is None: return
        self.bank.setdefault(fid, collections.deque(
            maxlen=self.bank_size)).append(hist)

    def distance_to_id(self, hist, fid):
        arr = self.bank.get(fid)
        if not arr or len(arr)==0: return 1.0
        mean_hist = np.mean(np.stack(arr, 0), axis=0)
        return bhattacharyya(hist, mean_hist)

    def best_id(self, hist):
        best_f, best_d = None, 1.0
        for fid in self.id_pool:
            d = self.distance_to_id(hist, fid)
            if d < best_d: best_d, best_f = d, fid
        return best_f, best_d


# ============================================================
#  [DYTB Má»I] Trajectory Interpolation
#  Khi track bá»‹ máº¥t 1..INTERP_MAX_GAP frame â†’ ná»™i suy tuyáº¿n tĂ­nh
#  giĂºp giáº£m ID switch do occlusion ngáº¯n (theo Pig-ByteTrack)
# ============================================================
class TrajectoryInterpolator:
    """
    LÆ°u lá»‹ch sá»­ bbox cá»§a má»—i fixed-ID.
    Khi 1 ID biáº¿n máº¥t táº¡m thá»i, ná»™i suy Ä‘á»ƒ váº«n váº½ bbox má».
    Khi ID xuáº¥t hiá»‡n láº¡i, reset.
    """
    def __init__(self, id_pool, max_gap=10):
        self.max_gap   = max_gap
        # last_seen[fid] = (frame_idx, box)
        self.last_seen = {fid: None for fid in id_pool}
        # gap_count[fid] = sá»‘ frame liĂªn tiáº¿p chÆ°a tháº¥y
        self.gap       = {fid: 0 for fid in id_pool}

    def update(self, fid, frame_idx, box):
        """Gá»i khi fid Ä‘Æ°á»£c detect á»Ÿ frame nĂ y."""
        self.last_seen[fid] = (frame_idx, np.array(box, dtype=np.float32))
        self.gap[fid] = 0

    def get_interpolated(self, fid, frame_idx):
        """
        Náº¿u fid khĂ´ng xuáº¥t hiá»‡n á»Ÿ frame nĂ y nhÆ°ng gap <= max_gap
        â†’ tráº£ vá» (box_ná»™i_suy, alpha) Ä‘á»ƒ váº½ bbox má».
        alpha giáº£m dáº§n tá»« 1â†’0 theo gap.
        Tráº£ vá» (None, 0) náº¿u khĂ´ng ná»™i suy Ä‘Æ°á»£c.
        """
        self.gap[fid] += 1
        g = self.gap[fid]
        if g > self.max_gap: return None, 0.0
        prev = self.last_seen.get(fid)
        if prev is None: return None, 0.0
        # Giá»¯ nguyĂªn tá»a Ä‘á»™ cuá»‘i (track tháº³ng â†’ khĂ´ng dá»± Ä‘oĂ¡n hÆ°á»›ng)
        alpha = max(0.0, 1.0 - g / self.max_gap)
        return prev[1].copy(), alpha


# ============================================================
#  [DYTB Má»I] 3rd Association: Centre-Distance
#  Sau 2 vĂ²ng IoU cá»§a ByteTrack, thĂªm vĂ²ng 3 dĂ¹ng khoáº£ng cĂ¡ch
#  tĂ¢m bbox Ä‘á»ƒ ghĂ©p cĂ¡c track cĂ²n sĂ³t (irregular pig motion)
# ============================================================
def centre_of_box(box):
    """box = [x1,y1,x2,y2] â†’ (cx,cy)"""
    return ((box[0]+box[2])/2.0, (box[1]+box[3])/2.0)

def centre_distance_match(unmatched_fixed_ids, new_dets,
                          interp, frame_idx,
                          fixed_to_raw, raw_to_fixed,
                          thr=CENTRE_DIST_THR):
    """
    unmatched_fixed_ids: set cá»§a cĂ¡c fid chÆ°a Ä‘Æ°á»£c ghĂ©p á»Ÿ vĂ²ng 1-2
    new_dets: list of {rid, box, score, hist} chÆ°a cĂ³ fixed-ID
    Tráº£ vá» dict {rid: fid} gá»£i Ă½ ghĂ©p.
    """
    if not unmatched_fixed_ids or not new_dets: return {}
    um_fids = list(unmatched_fixed_ids)
    cost = np.full((len(new_dets), len(um_fids)), np.inf)
    for i, det in enumerate(new_dets):
        dc = centre_of_box(det["box"])
        for j, fid in enumerate(um_fids):
            prev = interp.last_seen.get(fid)
            if prev is None: continue
            pc = centre_of_box(prev[1])
            dist = math.hypot(dc[0]-pc[0], dc[1]-pc[1])
            if dist <= thr:
                cost[i, j] = dist
    if not np.isfinite(cost).any():
        return {}
    safe_cost = np.where(np.isfinite(cost), cost, thr + 1e6)
    ri, ci = linear_sum_assignment(safe_cost)
    result = {}
    for i, j in zip(ri, ci):
        if np.isfinite(cost[i,j]):
            result[new_dets[i]["rid"]] = um_fids[j]
    return result
```

## Cell 5 â€” ID Pool & Assignment

```python
id_bank   = IDBank(ID_POOL, ID_BANK_SIZE)
interp    = TrajectoryInterpolator(ID_POOL, INTERP_MAX_GAP)

free_ids      = set(ID_POOL)
raw_to_fixed  = {}   # raw ByteTrack id â†’ fixed id
fixed_to_raw  = {}   # fixed id â†’ raw ByteTrack id

def take_free_id():
    if free_ids:
        fid = min(free_ids)
        free_ids.remove(fid)
        return fid
    return None

def bind_fixed(fid, rid):
    if fid in fixed_to_raw and fixed_to_raw[fid] != rid:
        old = fixed_to_raw[fid]
        if old in raw_to_fixed: del raw_to_fixed[old]
    fixed_to_raw[fid] = rid
    raw_to_fixed[rid] = fid

def free_fixed_by_raw(rid):
    if rid in raw_to_fixed:
        fid = raw_to_fixed[rid]
        free_ids.add(fid)
        del raw_to_fixed[rid]
        if fid in fixed_to_raw: del fixed_to_raw[fid]

def reassign_ids_consistency(active_detections):
    """
    Tá»‘i Æ°u toĂ n cá»¥c: gĂ¡n raw-id â†’ fixed-id dĂ¹ng Hungarian
    trĂªn ma tráº­n Bhattacharyya distance.
    """
    if not active_detections: return
    rids = [d["rid"] for d in active_detections]
    n, m = len(rids), len(ID_POOL)
    C = np.zeros((n, m), np.float32)
    for i, det in enumerate(active_detections):
        hist = det["hist"]
        for j, fid in enumerate(ID_POOL):
            d   = id_bank.distance_to_id(hist, fid)
            pen = 0.15 if (fid in fixed_to_raw and
                           fixed_to_raw[fid] != det["rid"]) else 0.0
            C[i, j] = min(1.0, d + pen)
    r, c = linear_sum_assignment(C)
    taken = set()
    for i, j in zip(r, c):
        rid = rids[i]; fid = ID_POOL[j]
        if fid in taken: continue
        cur = raw_to_fixed.get(rid)
        if cur == fid:
            taken.add(fid); continue
        if cur is not None and cur in fixed_to_raw:
            del fixed_to_raw[cur]
        if fid in fixed_to_raw:
            other = fixed_to_raw[fid]
            if other in raw_to_fixed: del raw_to_fixed[other]
        bind_fixed(fid, rid)
        if fid in free_ids: free_ids.discard(fid)
        taken.add(fid)
```

## Cell 6 â€” Time Parser

```python
time_lines = None
time_header = None

if os.path.exists(TIME_FILE):
    with open(TIME_FILE, "r", encoding="utf-8") as f:
        lines = [ln.strip() for ln in f if ln.strip()]
    if lines:
        time_lines = lines
        head = " ".join(lines)
        m = re.search(
            r"start\s*=\s*([0-9:\-\s\.TZ\+]+).*fps\s*=\s*([0-9\.]+)",
            head, re.IGNORECASE)
        if m:
            try:
                start = m.group(1).strip()
                fps_s = float(m.group(2))
                base  = datetime.datetime.fromisoformat(
                    start.replace("Z", "+00:00") if "T" in start else start)
                time_header = ("header", base, fps_s)
            except:
                time_header = None

def format_ts(idx):
    if time_lines is None: return f"frame {idx}"
    if time_header is None:
        return time_lines[idx-1] if len(time_lines) >= idx else f"frame {idx}"
    _, base, fps_s = time_header
    delta = datetime.timedelta(seconds=(idx-1)/fps_s)
    return (base + delta).isoformat(sep=" ", timespec="milliseconds")
```

## Cell 7 â€” Main Loop (DYTB)

```python
import gc
from collections import Counter
import ultralytics
from ultralytics import YOLO
print("Ultralytics version:", ultralytics.__version__)

# Video info
cap0 = cv2.VideoCapture(INPUT_VIDEO)
assert cap0.isOpened(), f"Cannot open: {INPUT_VIDEO}"
fps = cap0.get(cv2.CAP_PROP_FPS) or 30.0
W0  = int(cap0.get(cv2.CAP_PROP_FRAME_WIDTH))
H0  = int(cap0.get(cv2.CAP_PROP_FRAME_HEIGHT))
N0  = int(cap0.get(cv2.CAP_PROP_FRAME_COUNT) or 0)
cap0.release()

if not os.path.exists(BYTE_TRACK_YAML):
    print(f"[WARN] Missing {BYTE_TRACK_YAML}; fallback sang bytetrack.yaml m?c ??nh")
    BYTE_TRACK_YAML = 'bytetrack.yaml'

OUTPUT_FPS = fps / max(1, VID_STRIDE)

# Writer ch? t?o khi th?t s? c?n, v? ghi video full HD r?t ch?m.
writer = None
out_W, out_H = W0, H0
if WRITE_VIDEO:
    out_W = max(2, int(W0 * OUTPUT_SCALE))
    out_H = max(2, int(H0 * OUTPUT_SCALE))
    fourcc = cv2.VideoWriter_fourcc(*"mp4v")
    writer = cv2.VideoWriter(OUTPUT_VIDEO, fourcc, OUTPUT_FPS, (out_W, out_H))
    assert writer.isOpened(), f"Cannot write: {OUTPUT_VIDEO}"

mask  = load_mask(MASK_PATH, W0, H0) if USE_ROI else None
model = YOLO(WEIGHTS)
try:
    import torch
    torch.set_num_threads(1)
except Exception:
    pass

# Re-init tracking state
id_bank  = IDBank(ID_POOL, ID_BANK_SIZE)
interp   = TrajectoryInterpolator(ID_POOL, INTERP_MAX_GAP)
free_ids     = set(ID_POOL)
raw_to_fixed = {}
fixed_to_raw = {}

# Logs: JSONL streaming ?? kh?ng gi? to?n b? video trong RAM.
tracking_log = {} if STORE_IN_MEMORY_LOG else None
behavior_log = [] if STORE_IN_MEMORY_LOG else None
behavior_counts = Counter()
track_f = open(TRACK_JSONL, "w", encoding="utf-8")
beh_f = open(BEHAVIOR_JSONL, "w", encoding="utf-8")

def run_stream(yaml_path):
    return model.track(
        source=INPUT_VIDEO,
        tracker=yaml_path,
        conf=CONF_THRESH,
        iou=NMS_IOU,
        imgsz=IMGSZ,
        vid_stride=max(1, VID_STRIDE),
        stream=True,
        persist=True,
        verbose=False,
    )

processed = 0
last_frame_idx = 0
est_total = None
if N0 and MAX_FRAMES is not None:
    est_total = min(MAX_FRAMES, math.ceil(N0 / max(1, VID_STRIDE)))
elif N0:
    est_total = math.ceil(N0 / max(1, VID_STRIDE))

pbar = tqdm(desc="DYTB Tracking", total=est_total)
results = None
try:
    results = run_stream(BYTE_TRACK_YAML)
    for res in results:
        processed += 1
        frame_idx = 1 + (processed - 1) * max(1, VID_STRIDE)
        last_frame_idx = frame_idx
        frame = res.orig_img
        Hc, Wc = frame.shape[:2]
        if mask is None or mask.shape[:2] != (Hc, Wc):
            mask = load_mask(MASK_PATH, Wc, Hc) if USE_ROI else None

        dets = []
        if res.boxes is not None and len(res.boxes) > 0:
            xyxys = res.boxes.xyxy.cpu().numpy()
            confs = res.boxes.conf.cpu().numpy()
            ids   = res.boxes.id
            ids   = ids.cpu().numpy().astype(int) if ids is not None else None
            clss  = res.boxes.cls.cpu().numpy() if res.boxes.cls is not None else None

            if ids is not None:
                for i, box in enumerate(xyxys):
                    rid   = int(ids[i])
                    score = float(confs[i])
                    cls_i = int(clss[i]) if clss is not None else 0
                    if ALLOWED_CLASS is not None and cls_i != int(ALLOWED_CLASS):
                        continue
                    if USE_ROI and not roi_keep(mask, box):
                        continue
                    dets.append({"rid": rid, "box": box, "score": score, "cls": cls_i})

        for d in dets:
            d["hist"] = extract_hist_hsv(frame, d["box"])

        new_dets_no_fid = []
        for d in dets:
            if d["rid"] not in raw_to_fixed:
                best_id, dist = id_bank.best_id(d["hist"])
                if best_id is not None and dist <= REID_THR and best_id in free_ids:
                    bind_fixed(best_id, d["rid"])
                    free_ids.discard(best_id)
                else:
                    new_dets_no_fid.append(d)

        active_fids_now = set(raw_to_fixed.get(d["rid"]) for d in dets if d["rid"] in raw_to_fixed)
        active_fids_now.discard(None)
        candidate_fids  = set(fid for fid in ID_POOL if interp.last_seen[fid] is not None and fid not in active_fids_now)
        cd_matches = centre_distance_match(candidate_fids, new_dets_no_fid, interp, frame_idx, fixed_to_raw, raw_to_fixed)
        for rid, fid in cd_matches.items():
            bind_fixed(fid, rid)
            free_ids.discard(fid)
            new_dets_no_fid = [d for d in new_dets_no_fid if d["rid"] != rid]

        for d in new_dets_no_fid:
            if d["rid"] not in raw_to_fixed:
                fid = take_free_id()
                if fid is not None:
                    bind_fixed(fid, d["rid"])

        active_for_opt = [{"rid": d["rid"], "hist": d["hist"]} for d in dets if d["rid"] in raw_to_fixed]
        if ID_STEAL_FIX:
            reassign_ids_consistency(active_for_opt)

        seen_fids_this_frame = set()
        frame_tracks = []
        for d in dets:
            fid = raw_to_fixed.get(d["rid"])
            if fid is None:
                continue
            id_bank.update(fid, d["hist"])
            interp.update(fid, frame_idx, d["box"])
            seen_fids_this_frame.add(fid)
            x1, y1, x2, y2 = d["box"]
            rec = {
                "frame_idx": int(frame_idx),
                "fid": int(fid),
                "box": [float(x1), float(y1), float(x2), float(y2)],
                "score": float(d["score"]),
                "cls": int(d["cls"]),
            }
            frame_tracks.append(rec)
            track_f.write(json.dumps(rec, ensure_ascii=False) + "\n")

            bname = BEHAVIOR_MAP.get(d["cls"], "pig")
            behavior_counts[(fid, bname)] += 1
            beh_rec = {"frame_idx": int(frame_idx), "fid": int(fid), "behavior": bname}
            beh_f.write(json.dumps(beh_rec, ensure_ascii=False) + "\n")
            if STORE_IN_MEMORY_LOG:
                behavior_log.append((frame_idx, fid, bname))

        if STORE_IN_MEMORY_LOG and frame_tracks:
            tracking_log[frame_idx] = [
                {"fid": r["fid"], "box": r["box"], "score": r["score"], "cls": r["cls"]}
                for r in frame_tracks
            ]

        if WRITE_VIDEO:
            vis = apply_visual_mask(frame, mask) if USE_ROI else frame.copy()
            cv2.putText(vis, format_ts(frame_idx), (12, 28), cv2.FONT_HERSHEY_SIMPLEX, 0.8, (255,255,255), 2, cv2.LINE_AA)
            for d in dets:
                fid = raw_to_fixed.get(d["rid"])
                label = fid if fid is not None else d["rid"]
                x1, y1, x2, y2 = map(int, d["box"])
                color = (37 * label % 255, 97 * label % 255, 173 * label % 255)
                cv2.rectangle(vis, (x1, y1), (x2, y2), color, DRAW_THICKNESS)
                cv2.putText(vis, f"ID{label}", (x1, max(0, y1 - 6)), cv2.FONT_HERSHEY_SIMPLEX, FONT_SCALE, color, 2, cv2.LINE_AA)
            for fid in ID_POOL:
                if fid in seen_fids_this_frame:
                    continue
                ibox, alpha = interp.get_interpolated(fid, frame_idx)
                if ibox is None or alpha <= 0.0:
                    continue
                x1, y1, x2, y2 = map(int, ibox)
                color = (37 * fid % 255, 97 * fid % 255, 173 * fid % 255)
                cv2.rectangle(vis, (x1, y1), (x2, y2), color, 1)
                cv2.putText(vis, f"ID{fid}?", (x1, max(0, y1 - 6)), cv2.FONT_HERSHEY_SIMPLEX, FONT_SCALE * 0.8, color, 1, cv2.LINE_AA)
            if (out_W, out_H) != (Wc, Hc):
                vis = cv2.resize(vis, (out_W, out_H), interpolation=cv2.INTER_AREA)
            writer.write(vis)
        else:
            # V?n c?p nh?t gap n?i suy khi kh?ng v?, nh?ng kh?ng t?o b?n copy frame.
            for fid in ID_POOL:
                if fid not in seen_fids_this_frame:
                    interp.get_interpolated(fid, frame_idx)

        pbar.update(1)
        if MAX_FRAMES is not None and processed >= MAX_FRAMES:
            break
        if processed % 200 == 0:
            track_f.flush(); beh_f.flush(); gc.collect()
finally:
    pbar.close()
    if results is not None and hasattr(results, "close"):
        results.close()
    track_f.close()
    beh_f.close()
    if writer is not None:
        writer.release()
    gc.collect()

print(f"[OK] Processed frames: {processed:,} (last source frame: {last_frame_idx:,}/{N0:,})")
print(f"[OK] Input FPS: {fps:.3f} | Output FPS: {OUTPUT_FPS:.3f} | VID_STRIDE: {max(1, VID_STRIDE)}")
print(f"[OK] Tracks JSONL -> {TRACK_JSONL}")
print(f"[OK] Behavior JSONL -> {BEHAVIOR_JSONL}")
if WRITE_VIDEO:
    print(f"[OK] Video -> {OUTPUT_VIDEO}")
```

## Cell 9 â€” MOT Metrics: HOTA / MOTA / IDF1 / MOTP / IDs / FP / FN

```python
import numpy as np
from scipy.optimize import linear_sum_assignment
from collections import defaultdict
import os, json, math

GT_FILE    = '/content/gt.txt'
PRED_FILE  = '/content/pred.txt'
IOU_THRESH = 0.50

# â”€â”€ Export pred.txt tu tracking loop â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
def export_mot_predictions(tracking_results_per_frame, out_path):
    lines = []
    for fidx in sorted(tracking_results_per_frame.keys()):
        for det in tracking_results_per_frame[fidx]:
            x1,y1,x2,y2 = det['box']
            w = x2-x1; h = y2-y1
            tid  = det['fid']
            conf = det.get('score', 1.0)
            lines.append(f'{fidx},{tid},{x1:.1f},{y1:.1f},{w:.1f},{h:.1f},{conf:.3f},-1,-1,-1')
    with open(out_path, 'w') as f:
        f.write('\n'.join(lines))
    print(f'[OK] Pred -> {out_path}  ({len(lines)} dong)')

# â”€â”€ Doc file MOT â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
def read_mot_file(path):
    data = defaultdict(list)
    if not os.path.exists(path): return data
    with open(path) as f:
        for line in f:
            line = line.strip()
            if not line or line.startswith('#'): continue
            parts = line.split(',')
            if len(parts) < 6: continue
            fid = int(float(parts[0])); tid = int(float(parts[1]))
            x=float(parts[2]); y=float(parts[3])
            w=float(parts[4]); h=float(parts[5])
            conf = float(parts[6]) if len(parts)>6 else 1.0
            if w<=0 or h<=0: continue
            data[fid].append((tid, x, y, x+w, y+h, conf))
    return data

# â”€â”€ IoU â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
def iou_box(b1, b2):
    ix1=max(b1[0],b2[0]); iy1=max(b1[1],b2[1])
    ix2=min(b1[2],b2[2]); iy2=min(b1[3],b2[3])
    iw=max(0,ix2-ix1); ih=max(0,iy2-iy1)
    inter=iw*ih
    a1=(b1[2]-b1[0])*(b1[3]-b1[1]); a2=(b2[2]-b2[0])*(b2[3]-b2[1])
    union=a1+a2-inter
    return inter/union if union>0 else 0.0

# â”€â”€ CLEAR MOT: MOTA / MOTP / IDs / FP / FN / TP â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
def compute_clearmot(gt_data, pred_data, iou_thr=0.5):
    all_frames=sorted(set(list(gt_data.keys())+list(pred_data.keys())))
    total_gt=0; total_tp=0; total_fp=0; total_fn=0; total_idsw=0
    total_motp=0.0; total_tp_motp=0
    gt_to_pred_last={}
    for frame in all_frames:
        gts=gt_data.get(frame,[]); preds=pred_data.get(frame,[])
        total_gt+=len(gts)
        if not gts:  total_fp+=len(preds); continue
        if not preds: total_fn+=len(gts);  continue
        cost=np.zeros((len(gts),len(preds)))
        for i,g in enumerate(gts):
            for j,p in enumerate(preds):
                cost[i,j]=iou_box((g[1],g[2],g[3],g[4]),(p[1],p[2],p[3],p[4]))
        ri,ci=linear_sum_assignment(-cost)
        mgt=set(); mpr=set()
        for i,j in zip(ri,ci):
            if cost[i,j]>=iou_thr:
                gid=gts[i][0]; pid=preds[j][0]
                mgt.add(i); mpr.add(j)
                total_tp+=1; total_motp+=cost[i,j]; total_tp_motp+=1
                prev=gt_to_pred_last.get(gid)
                if prev is not None and prev!=pid: total_idsw+=1
                gt_to_pred_last[gid]=pid
        total_fn+=len(gts)-len(mgt); total_fp+=len(preds)-len(mpr)
    mota=1.0-(total_fn+total_fp+total_idsw)/max(1,total_gt)
    motp=total_motp/total_tp_motp if total_tp_motp>0 else 0.0
    prec=total_tp/max(1,total_tp+total_fp)
    rec=total_tp/max(1,total_gt)
    return {'MOTA':round(mota*100,2),'MOTP_IoU':round(motp*100,2),
            'Precision':round(prec*100,2),'Recall':round(rec*100,2),
            'TP':total_tp,'FP':total_fp,'FN':total_fn,'IDs':total_idsw,'GT':total_gt}

# â”€â”€ IDF1 â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
def compute_idf1(gt_data, pred_data, iou_thr=0.5):
    all_frames=sorted(set(list(gt_data.keys())+list(pred_data.keys())))
    gt_traj=defaultdict(list); pred_traj=defaultdict(list)
    frame_matches={}
    for frame in all_frames:
        gts=gt_data.get(frame,[]); preds=pred_data.get(frame,[])
        for g in gts: gt_traj[g[0]].append((frame,(g[1],g[2],g[3],g[4])))
        for p in preds: pred_traj[p[0]].append((frame,(p[1],p[2],p[3],p[4])))
        if not gts or not preds: frame_matches[frame]=[]; continue
        cost=np.zeros((len(gts),len(preds)))
        for i,g in enumerate(gts):
            for j,p in enumerate(preds):
                cost[i,j]=iou_box((g[1],g[2],g[3],g[4]),(p[1],p[2],p[3],p[4]))
        ri,ci=linear_sum_assignment(-cost)
        frame_matches[frame]=[(gts[i][0],preds[j][0],cost[i,j])
                              for i,j in zip(ri,ci) if cost[i,j]>=iou_thr]
    gt_ids=sorted(gt_traj.keys()); pred_ids=sorted(pred_traj.keys())
    if not gt_ids or not pred_ids: return {'IDF1':0.0,'IDTP':0,'IDFP':0,'IDFN':0}
    overlap=defaultdict(int)
    for frame,matches in frame_matches.items():
        for gid,pid,_ in matches: overlap[(gid,pid)]+=1
    cost_mat=np.zeros((len(gt_ids),len(pred_ids)))
    for i,gid in enumerate(gt_ids):
        for j,pid in enumerate(pred_ids):
            cost_mat[i,j]=overlap.get((gid,pid),0)
    ri,ci=linear_sum_assignment(-cost_mat)
    idtp=sum(int(cost_mat[i,j]) for i,j in zip(ri,ci))
    total_gt_det=sum(len(v) for v in gt_traj.values())
    total_pred_det=sum(len(v) for v in pred_traj.values())
    idfp=total_pred_det-idtp; idfn=total_gt_det-idtp
    idf1=2*idtp/max(1,2*idtp+idfp+idfn)
    return {'IDF1':round(idf1*100,2),'IDTP':idtp,'IDFP':idfp,'IDFN':idfn}

# â”€â”€ HOTA = sqrt(DetA * AssA) â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
def compute_hota(gt_data, pred_data, iou_thr=0.5):
    all_frames=sorted(set(list(gt_data.keys())+list(pred_data.keys())))
    frame_tp_pairs=[]; total_tp=0; total_fp=0; total_fn=0
    for frame in all_frames:
        gts=gt_data.get(frame,[]); preds=pred_data.get(frame,[])
        if not gts: total_fp+=len(preds); continue
        if not preds: total_fn+=len(gts); continue
        cost=np.zeros((len(gts),len(preds)))
        for i,g in enumerate(gts):
            for j,p in enumerate(preds):
                cost[i,j]=iou_box((g[1],g[2],g[3],g[4]),(p[1],p[2],p[3],p[4]))
        ri,ci=linear_sum_assignment(-cost)
        mgt=set(); mpr=set()
        for i,j in zip(ri,ci):
            if cost[i,j]>=iou_thr:
                total_tp+=1; mgt.add(i); mpr.add(j)
                frame_tp_pairs.append((gts[i][0],preds[j][0]))
        total_fn+=len(gts)-len(mgt); total_fp+=len(preds)-len(mpr)
    det_a=total_tp/max(1,total_tp+total_fp+total_fn)
    if not frame_tp_pairs: return {'HOTA':0.0,'DetA':0.0,'AssA':0.0}
    pair_count=defaultdict(int); gt_count=defaultdict(int); pred_count=defaultdict(int)
    for gid,pid in frame_tp_pairs:
        pair_count[(gid,pid)]+=1; gt_count[gid]+=1; pred_count[pid]+=1
    ass=[tpa/max(1,tpa+(pred_count[pid]-tpa)+(gt_count[gid]-tpa))
         for (gid,pid),tpa in pair_count.items()]
    ass_a=float(np.mean(ass)) if ass else 0.0
    hota=math.sqrt(det_a*ass_a)
    return {'HOTA':round(hota*100,2),'DetA':round(det_a*100,2),'AssA':round(ass_a*100,2)}

# â”€â”€ Per-frame ID switch (cho timeline) â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
def compute_per_frame_idsw(gt_data, pred_data, iou_thr=0.5):
    all_frames=sorted(set(list(gt_data.keys())+list(pred_data.keys())))
    n_frames=max(all_frames)+1 if all_frames else 1
    pf=np.zeros(n_frames,dtype=int)
    gt_to_pred_last={}
    for frame in all_frames:
        gts=gt_data.get(frame,[]); preds=pred_data.get(frame,[])
        if not gts or not preds: continue
        cost=np.zeros((len(gts),len(preds)))
        for i,g in enumerate(gts):
            for j,p in enumerate(preds):
                cost[i,j]=iou_box((g[1],g[2],g[3],g[4]),(p[1],p[2],p[3],p[4]))
        ri,ci=linear_sum_assignment(-cost)
        for i,j in zip(ri,ci):
            if cost[i,j]>=iou_thr:
                gid=gts[i][0]; pid=preds[j][0]
                prev=gt_to_pred_last.get(gid)
                if prev is not None and prev!=pid: pf[frame]+=1
                gt_to_pred_last[gid]=pid
    return pf

# â”€â”€ CHAY TOAN BO â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
print('='*55)
print('  MOT METRICS EVALUATION â€” PIG TRACKING (DYTB)')
print('='*55)

if not os.path.exists(GT_FILE):
    print(f"""
[!] Chua co ground truth: {GT_FILE}

Cach tao GT file (MOT format):
  Moi dong: frame_id,track_id,x,y,w,h,1,-1,-1,-1
  Vi du:    1,1,120,80,60,90,1,-1,-1,-1
            2,1,122,82,60,90,1,-1,-1,-1

Cong cu annotate: CVAT, DarkLabel, LabelImg
-> Xuat sang MOT format -> luu vao {GT_FILE}

Chay Cell 10 de xem demo bieu do (du lieu mau).
""")
elif not os.path.exists(PRED_FILE):
    print(f'[!] Chua co pred file: {PRED_FILE}')
    print('    Chay tracking loop (Cell 7) truoc, sau do goi:')
    print('    export_mot_predictions(tracking_results, PRED_FILE)')
else:
    gt_data  = read_mot_file(GT_FILE)
    pd_data  = read_mot_file(PRED_FILE)
    clear    = compute_clearmot(gt_data, pd_data, IOU_THRESH)
    idf1_r   = compute_idf1(gt_data, pd_data, IOU_THRESH)
    hota_r   = compute_hota(gt_data, pd_data, IOU_THRESH)
    pf_ids   = compute_per_frame_idsw(gt_data, pd_data, IOU_THRESH)
    metrics  = {**hota_r, **idf1_r, **clear}
    out_json = PRED_FILE.replace('.txt','_metrics.json')
    with open(out_json,'w') as f:
        json.dump({'metrics':metrics,'per_frame_idsw':pf_ids.tolist()},f,indent=2)
    rows=[
        ('HOTA','%  cao hon tot hon'),('DetA','%  detection accuracy'),
        ('AssA','%  association accuracy'),('IDF1','%  ID consistency'),
        ('MOTA','%  overall accuracy'),('MOTP_IoU','%  localisation accuracy'),
        ('Precision','%'),('Recall','%'),
        ('IDs','   it hon tot hon'),('FP','   False Positive'),
        ('FN','   False Negative'),('TP','   True Positive'),('GT','   Ground truth'),
    ]
    print(f"{'Metric':<14} {'Value':>8}   Note")
    print('-'*45)
    for k,note in rows:
        v=metrics.get(k,'â€”')
        print(f'  {k:<14} {str(v):>8}   {note}')
    print(f'\n[OK] Metrics saved -> {out_json}')
    print('-> Chay Cell 10 de xem bieu do')
```
